```python
from pathlib import Path

import torch
from turyn_policy_transformer import PolicyOnlyTransformer, evaluate_model_with_sat


def evaluate_all_models_with_sat(models_dir="models", r=9):
    """Run SAT evaluation for every checkpoint in models_dir recursively."""
    model_paths = sorted(Path(models_dir).rglob("*.pt"))

    if not model_paths:
        print(f"No model checkpoints found in {models_dir}")
        return

    for model_path in model_paths:
        print(f"\nEvaluating: {model_path}")
        model = PolicyOnlyTransformer()
        state_dict = torch.load(model_path, map_location="cpu")
        model.load_state_dict(state_dict["state_dict"])
        model_sat_results = evaluate_model_with_sat(model, r=r)
        print("\n--- SAT Evaluation (model policy) ---")
        print(f"  count_right:     {model_sat_results['count_right']}")
        print(f"  count_wrong:     {model_sat_results['count_wrong']}")
        print(f"  first_incorrect: {model_sat_results['first_incorrect']}")
        print(f"  num_unsat:       {model_sat_results['num_unsat']}")
        print(f"  sum_unsat:       {model_sat_results['sum_unsat']}")


evaluate_all_models_with_sat(models_dir="models", r=9)
```
